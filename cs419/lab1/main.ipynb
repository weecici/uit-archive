```python
text = """experimental investigation of the aerodynamics of a wing in a slipstream . an experimental study of a wing in a propeller slipstream was made in order to determine the spanwise distribution of the lift increase due to slipstream at different angles of attack of the wing and at different free stream to slipstream velocity ratios .  the results were intended in part as an evaluation basis for different theoretical treatments of this problem . the comparative span loading curves, together with supporting evidence, showed that a substantial part of the lift increment produced by the slipstream was due to a /destalling/ or boundary layer control effect .  the integrated remaining lift increment, after subtracting this destalling lift, was found to agree well with a potential flow theory . an empirical evaluation of the destalling effects was made for the specific configuration of the experiment . """
sents = sent_tokenize(text)
print(sents)
```

```python
words_sent_0 = word_tokenize(sents[0])
print(words_sent_0)
```

```python
stoplist = stopwords.words("english")
print(stoplist)
print(len(stoplist))
```

```python
filtered_words_sent_0 = []
for word in words_sent_0:
    if word in stoplist:
        continue
    filtered_words_sent_0.append(word)
print(filtered_words_sent_0)
```

```python
tokens = []
for word in filtered_words_sent_0:
    if word.isalnum():
        tokens.append(word)
print(tokens)
```

```python
stemmer = PorterStemmer()
terms = []
for word in tokens:
    terms.append(stemmer.stem(word))
print(terms)
```

**VÍ DỤ**

Cho tập tài liệu D gồm:

d1: Romeo and Juliet

d2: Juliet: Oh happy dagger

d3: Romeo died by dagger

d4: "live free or die" is from New-Hampshire

d5: He is from New-Hampshire

Yêu cầu:

- Lập chỉ mục cho tập D theo mô hình vector

- Xử lý truy vấn q trên chỉ mục vừa lập. Biết truy vấn q là happy dagger


```python
docs = [
    "Romeo and Juliet",
    "Juliet: Oh happy dagger",
    "Romeo died by dagger",
    '"live free or die" is from New-Hampshire',
    "He is from New-Hampshire",
]
query = "happy dagger"
```

```python
puncts = [".", ",", ":", "`", '"', "'", "!", "?", "``", "''"]
stoplist.append("oh")
```

```python
processed_docs = []
for i in range(len(docs)):
    doc_terms = []
    doc_sents = sent_tokenize(docs[i])
    for j in range(len(doc_sents)):
        tokens = word_tokenize(doc_sents[j])
        for token in tokens:
            token = token.lower()
            if (
                token.isdigit()
                or token.isnumeric()
                or token in puncts
                or token in stoplist
            ):
                continue
            doc_terms.append(stemmer.stem(token))
    processed_docs.append(doc_terms)
print(processed_docs)
```

```python
from nltk.probability import FreqDist

all_terms = []
for doc in processed_docs:
    all_terms.extend(doc)
freq_dist = FreqDist(all_terms)
for term, freq in freq_dist.items():
    print(term, freq)
```

```python
terms = {term: i for i, term in enumerate(list(freq_dist.keys()))}
term_docs = [[0] * len(processed_docs) for _ in range(len(terms))]

for i in range(len(processed_docs)):
    for w in processed_docs[i]:
        term_docs[terms[w]][i] += 1
print(term_docs)
```

**VÍ DỤ**

Mở chỉ mục được lưu trong thư mục ind và xử lý truy vấn "happy dagger"


```python
from whoosh import qparser, scoring
import whoosh.index as index

idx = index.open_dir("example_idx")
searcher = idx.searcher(weighting=scoring.TF_IDF())

parser = qparser.QueryParser("content", idx.schema, group=qparser.OrGroup)
query = parser.parse("happy dagger")
results = searcher.search(query, limit=None)
for i, res in enumerate(results):
    print(i, res)
```

```python
!mkdir -p cranfield_idx
```

```python
import os

base_dir = "Cranfield/docs"
doc_paths = [os.path.join(base_dir, path) for path in os.listdir(base_dir)]
```

```python
schema = Schema(doc_id=STORED(), content=TEXT(analyzer=StandardAnalyzer(), stored=True))
idx = create_in("cranfield_idx", schema=schema)
writer = idx.writer()

for i, path in enumerate(doc_paths):
    with open(path, "r") as f:
        writer.add_document(doc_id=f"{i+1}", content=f.read())
writer.commit()
```

```python
import re

queries_path = "Cranfield/test/query.txt"
queries = []
with open(queries_path, "r") as f:
    raw_queries = f.readlines()
    for raw_query in raw_queries:
        query = re.sub(r"^\d+\t", "", raw_query)
        queries.append(query)
print(len(queries))
```

```python
idx = index.open_dir("cranfield_idx")
searcher = idx.searcher(weighting=scoring.TF_IDF())
parser = qparser.QueryParser("content", idx.schema, group=qparser.OrGroup)
results = []

for query in queries:
    query = parser.parse(query)
    matches = searcher.search(query, limit=None)
    results.append([])
    for match in matches:
        results[-1].append(match["doc_id"])

for i in range(10):
    print(f"{i + 1}: {len(results[i])}")
```
