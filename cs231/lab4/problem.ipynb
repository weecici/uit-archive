#Khai báo thư viện

```python
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt

from skimage import feature
from sklearn.model_selection import train_test_split

from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression


from sklearn.metrics import accuracy_score, classification_report
```

# Đọc dữ liệu

```python
# Hàm đọc dữ liệu từ thư mục folder
def load_images_from_folder(folder):
    images = []
    labels = []
    class_names = sorted(os.listdir(folder))

    for label, class_name in enumerate(class_names):
        class_folder = os.path.join(folder, class_name)
        if os.path.isdir(class_folder):
            for filename in os.listdir(class_folder):
                img_path = os.path.join(class_folder, filename)
                img = cv2.imread(img_path)
                if img is not None:
                    images.append(img)
                    labels.append(label)

    return images, labels, class_names
```

```python
# Đọc dữ liệu từ thư mục Dataset
dir_dataset = "/content/drive/MyDrive/CS231/Dataset/MyPedestrian"


pedestrian_path = os.path.join(dir_dataset, 'Pedestrian')
pedestrian = os.listdir(pedestrian_path)

non_pedestrian_path = os.path.join(dir_dataset, 'NonPedestrian')
non_pedestrian = os.listdir(non_pedestrian_path)

labels = ['pedestrian']*len(pedestrian) + ['non_pedestrian']*len(non_pedestrian)
y_labels = np.array(labels)
```

```python
def load_data(pedestrian,pedestrian_path, non_pedestrian, non_pedestrian_path):
    data=[]
    for filename in pedestrian:
      img_path=os.path.join(pedestrian_path, filename)
      img = cv2.imread(img_path, 0)
      img = cv2.resize(img, (128, 256))
      H = feature.hog(img, orientations=9, pixels_per_cell=(8, 8), cells_per_block=(2, 2),
                                    block_norm='L2-Hys', visualize=False, feature_vector =True )
      data.append(H)
    for filename in non_pedestrian:
      img_path=os.path.join(non_pedestrian_path, filename)
      img = cv2.imread(img_path, 0)
      img = cv2.resize(img, (128, 256))
      H = feature.hog(img, orientations=9, pixels_per_cell=(8, 8), cells_per_block=(2, 2),
                                    block_norm='L2-Hys', visualize=False, feature_vector =True )
      data.append(H)
    return np.array(data)
```

```python
data = load_data(pedestrian,pedestrian_path, non_pedestrian, non_pedestrian_path)
print(f"Tổng số ảnh : {len(data)}")
```

```python
# Thực hiện chia dữ liệu
X_train, X_test, y_train, y_test = train_test_split (data, labels, test_size=0.3, train_size=0.7, random_state=42)
```

```python
X_train = np.array(X_train)
X_test = np.array(X_test)
y_train = np.array(y_train)
y_test = np.array(y_test)

print(f"Kích thước tập train: {X_train.shape}, {y_train.shape}")
print(f"Kích thước tập test: {X_test.shape}, {y_test.shape}")
```

```python
knn= KNeighborsClassifier ()
knn.fit(X_train, y_train);
```

```python
test_pred = knn.predict(X_test)
print(classification_report(y_test, test_pred))

acc_knn = accuracy_score(y_test, test_pred)
print(f"Độ chính xác với KNeighborsClassifier: {acc_knn:.4f}")
```

```python
Logistic = LogisticRegression()
Logistic.fit(X_train, y_train);
```

```python
test_pred = Logistic.predict(X_test)
print(classification_report(y_test, test_pred))

acc_logistic = accuracy_score(y_test, test_pred)
print(f"Độ chính xác với LogisticRegression: {acc_logistic:.4f}")
```
